## The reproducible analysis workflow

1. **Question:** What do we want to learn?
2. **Data:** Which observations and features can help answer it?
3. **Evidence:** What should we calculate and display to answer the question?
4. **Check:** Does the result make sense in context? Do a few original records support the meaning of the calculation?
5. **Conclusion:** What claim is supported?
6. **Limitation:** What should we avoid concluding?

```python
duration_df = music_df[["duration_ms"]].copy()
duration_df["over_four_minutes"] = duration_df["duration_ms"] > 240000
duration_df.head()
```

```python
long_count = duration_df["over_four_minutes"].sum()
total_count = len(duration_df)
long_percentage = long_count / total_count * 100

print("Entries longer than four minutes:", long_count)
print("Total track entries:", total_count)
print(f"Percentage: {long_percentage:.2f}%")
```

**Which specific value or comparison answers your question? Explain how your calculations connect the selected data to that result.**

Your answer:There are 39,180 track entries longer than four minutes out of 114,000 total entries. I counted entries with duration_ms greater than 240,000, divided that count by 114,000, and multiplied by 100. The result is approximately 34.37%.

```python
energy_df = music_df.loc[
    (music_df["energy"] >= 0.7) | (music_df["energy"] < 0.3),
    ["energy", "danceability"]
].copy()

energy_df["energy_group"] = "Low"
energy_df.loc[energy_df["energy"] >= 0.7, "energy_group"] = "High"

energy_df.head()
```

```python
energy_df.groupby("energy_group")["danceability"].agg(
    ["count", "mean"]
)
```

**Which specific value or comparison answers your question? Explain how your calculations connect the selected data to that result.**

Your answer:I grouped the selected entries by energy level and calculated average danceability for each group. The 54,649 high-energy entries averaged 0.572, while the 13,558 low-energy entries averaged 0.439. Average danceability was approximately 0.133 higher in the high-energy group on the 0–1 scale. Entries with energy scores from 0.3 up to, but not including, 0.7 were excluded.
